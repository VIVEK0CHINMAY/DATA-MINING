# DECISION TREE - CANCER DATASET

---------------

#### STEP 1: LOAD THE DATASET

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.metrics import accuracy_score,classification_report
```

```python
df=pd.read_csv("cancer_data.csv")
df.head()
```

```python
df.drop(['id','Unnamed: 32'],axis=1,inplace=True)
```

#### STEP 2: SPLIT THE FEATURES AND LABEL

```python
x=df.drop('diagnosis',axis=1)
y=df['diagnosis']
```

#### STEP 3: TRAINING AND TEST SPLIT

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

#### STEP 4: DEFINE THE CLASSIFICATION MODEL

```python
model=DecisionTreeClassifier(criterion='entropy',random_state=42)
model.fit(x_train,y_train)
```

#### STEP 5: PERFORMANCE EVALUATION

```python
y_pred=model.predict(x_test)
print(classification_report(y_test,y_pred))
```

#### STEP 6: CLASSIFY THE NEW SAMPLE DATA

```python
y_pred = model.predict(df.drop('diagnosis', axis=1).iloc[[0]])
print(y_pred)
```

#### STEP 7: VISULIZATION

```python
feature_names = df.columns.drop('diagnosis')
plt.figure(figsize=(12, 8))
plot_tree(model, filled=True, feature_names=feature_names, class_names= model.classes_)
plt.title("Decision Tree for cancer disease Classification")
plt.show()
```

# NAIVE BAYES

--------------

#### STEP 1: LOAD THE DATASET

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix,ConfusionMatrixDisplay
```

```python
df=pd.read_csv("fruits.csv")
df.head()
```

#### STEP 2: SPLIT THE FEATURES AND LABEL

```python
x=df[['weight','size','color_score']]
y=df['Fruit_label']
```

#### STEP 3: TRAINING AND TEST SPLIT

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

#### STEP 4: DEFINE THE CLASSIFICATION MODEL

```python
from sklearn.naive_bayes import GaussianNB
model=GaussianNB()
model.fit(x_train,y_train)
```

#### STEP 5: PERFORMANCE EVALUATION

```python
y_pred=model.predict(x_test)
print(classification_report(y_test,y_pred))
```

#### STEP 6: CLASSIFY THE NEW SAMPLE DATA

```python
y_pred=model.predict([[150,7.5,0.50]])
print(y_pred)
```

#### STEP 7: VISULIZATION

```python
cm=confusion_matrix(y_test,y_pred , labels=model.classes_)
disp=ConfusionMatrixDisplay(confusion_matrix=cm,display_labels=model.classes_)
disp.plot()
```

# NAVIE BAYES -NB USER DATASET

---

#### STEP 1: LOAD THE DATASET

```python
df=pd.read_csv("NBUser data.csv")
df.head()
```

#### STEP 2: SPLIT THE FEATURES AND LABEL

```python
df['Gender']=df['Gender'].map({'Male':1,'Female':0})
x=df[['Gender','Age','EstimatedSalary']]
y=df['Purchased']
```

#### STEP 3: TRAINING AND TEST SPLIT

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

#### STEP 4: DEFINE THE CLASSIFICATION MODEL

```python
from sklearn.naive_bayes import GaussianNB
model=GaussianNB()
model.fit(x_train,y_train)
```

#### STEP 5: PERFORMANCE EVALUATION

```python
y_pred=model.predict(x_test)
print(classification_report(y_test,y_pred))
```

#### STEP 6: CLASSIFY THE NEW SAMPLE DATA

```python
y_pred=model.predict([[1,19,19000]])
print(y_pred)
```

#### STEP 7: VISULIZATION

```python
cm=confusion_matrix(y_test,y_pred,labels=model.classes_)
disp=ConfusionMatrixDisplay(confusion_matrix=cm,display_labels=model.classes_)
disp.plot()
```
